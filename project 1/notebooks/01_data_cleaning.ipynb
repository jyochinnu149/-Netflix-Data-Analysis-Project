# Netflix Data Cleaning Notebook

This notebook prepares the raw Netflix-style dataset for analysis by removing duplicates, fixing missing values, standardizing fields, and creating useful derived columns.

```python
import numpy as np
import pandas as pd
from pathlib import Path

root = Path('..')
df = pd.read_csv(root / 'data' / 'netflix_titles_raw.csv')
print(df.head())
print('Rows before:', len(df))

df = df.drop_duplicates().copy()
df.columns = ['show_id', 'type', 'title', 'director', 'cast', 'country', 'date_added', 'release_year', 'rating', 'duration', 'listed_in', 'description']
for col in ['type', 'title', 'director', 'cast', 'country', 'rating', 'listed_in', 'description']:
    df[col] = df[col].fillna('Unknown').astype(str).str.strip()

df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['type'] = df['type'].str.title()
df['country'] = df['country'].str.replace(r'\s*,\s*', ', ', regex=True)
df['main_genre'] = df['listed_in'].fillna('General').str.split(',').str[0].str.strip().str.title()
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month_name()
df['content_age'] = pd.Timestamp.today().year - df['release_year']

df.to_csv(root / 'data' / 'netflix_titles_cleaned.csv', index=False)
print('Rows after cleaning:', len(df))
print(df.head())
```
